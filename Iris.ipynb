```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sqlite3

%matplotlib inline
```

```python
conn = sqlite3.connect(':memory:')
df = pd.read_csv('Iris.csv')
df.to_sql("Sight", conn, index = False, if_exists = 'replace')
query = (''' SELECT * FROM Sight''')
df = pd.read_sql_query(query, conn)
df.head()
```
