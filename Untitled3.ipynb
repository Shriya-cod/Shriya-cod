```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('8. Netflix Dataset.csv')
df.head()
```

```python
df.dropna()
```

```python
print(df.columns.to_list)
```

```python
type_counts = df['Category'].value_counts()
plt.bar(type_counts.index, type_counts.values, color = ['blue', 'red'])
plt.xlabel('Type')
plt.ylabel('Count')
plt.title('Number of movies VS TV shows')
plt.tight_layout()
plt.show()
```

```python
df.drop_duplicates()
```

```python
rating_counts = df['Rating'].value_counts()
plt.figure(figsize = (8, 5))
plt.pie(rating_counts, labels = rating_counts.index, autopct = '%1.1f%%')
plt.title('Percentage of content rating')
plt.tight_layout()
plt.show()
```

```python
df['Release'] = pd.to_datetime(df['Release_Date'], errors = 'coerce')
df['Year'] = df['Release'].dt.year
year_counts = df['Year'].value_counts().sort_index()
plt.scatter(year_counts.index, year_counts.values, color = 'red', marker = 'o')
plt.xlabel('Year')
plt.ylabel('Shows')
plt.title('Shows per year')
plt.grid(True)
plt.show()
```

```python
c = df['Country'].value_counts().head(10)
plt.barh(c.index, c.values, color = 'blue')
plt.xlabel('Shows')
plt.ylabel('Country')
plt.title('Top 10 countries')
plt.show()
```

```python
content_per_year = df.groupby(['Year', 'Category']).size().unstack().fillna(0)

fig, ax = plt.subplots(1, 2, figsize = (10, 3))

ax[0].plot(content_per_year.index, content_per_year['Movie'], color = 'blue')
ax[0].set_xlabel('Year')
ax[0].set_ylabel('Movies')

ax[1].plot(content_per_year.index, content_per_year['TV Show'], color = 'black')
ax[1].set_xlabel('Year')
ax[1].set_ylabel('TV Show')
fig.suptitle('Comparision of Movies & TV Shows per year')

plt.tight_layout()
plt.savefig('Comparision of Movies & TV Shows per year.jpg', dpi = 300, bbox_inches = 'tight')
plt.show()
```
