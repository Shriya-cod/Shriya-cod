```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
```

```python
sales = pd.read_csv('hw_200.csv')
sales.head()
```

```python
sales[' Height(Inches)"'].plot(kind = 'hist', figsize = (10,3))
```

```python
sales[' Height(Inches)"'].value_counts().plot(kind = 'pie', figsize = (3,3))
```

```python
sales[' Height(Inches)"'].value_counts()
```

```python
ax = sales[' Height(Inches)"'].value_counts().plot(kind = 'bar', figsize = (10,3))
ax
```

```python
corr = sales.corr()
corr
```

```python
corr = sales.corr(numeric_only=True)
fig = plt.figure(figsize = (3,3))
plt.matshow(corr, cmap = 'RdBu')
plt.xticks(range(len(corr.columns)), corr.columns, rotation = 'vertical')
plt.yticks(range(len(corr.columns)), corr.columns)
```

```python
sales.plot(kind = 'scatter', x = ' Height(Inches)"', y = 'Index', figsize = (10,3))
```

```python
# Box plot height by weight
# by X axis
# set by y label Y axis

ax = sales[[' Height(Inches)"', ' "Weight(Pounds)"' ]].boxplot(by = ' "Weight(Pounds)"', figsize = (10,3))
ax.set_ylabel(' Height(Inches)"')
```

```python
print(sales.columns.tolist())
```

```python
boxplot_cols = ['Index',' Height(Inches)"', ' "Weight(Pounds)"' ]
sales[boxplot_cols].plot(kind = 'box', subplots = True, layout = (2,3), figsize = (10,3))
```
